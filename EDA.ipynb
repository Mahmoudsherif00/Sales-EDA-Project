# What are the most frequent categories?

```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

top_n = 5

cols = 2
rows = math.ceil(len(categorical_features) / cols)

fig, axes = plt.subplots(rows, cols, figsize=(20, 8 * rows))
axes = axes.flatten()

colors = sns.color_palette("Set2", top_n)

for i, feature in enumerate(categorical_features):

    values = df[feature].value_counts().head(top_n)

    axes[i].pie(
        values,
        labels=values.index,
        autopct='%1.1f%%',
        colors=colors[:len(values)],
        startangle=90,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1}
    )

    axes[i].set_title(
        f'Top {top_n} - {feature}',
        fontsize=15,
        fontweight='bold'
    )

for j in range(len(categorical_features), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
correlation = df[Numerical_features].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True
)

plt.title('Correlation Matrix', fontsize=16, fontweight='bold')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)

    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    r, k = confusion_matrix.shape

    return np.sqrt(chi2 / (n * min(r - 1, k - 1)))

correlation = pd.DataFrame(
    index=categorical_features,
    columns=categorical_features,
    dtype=float
)

for col1 in categorical_features:
    for col2 in categorical_features:
        correlation.loc[col1, col2] = cramers_v(
            df[col1], df[col2]
        )

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=0,
    vmax=1
)

plt.title('Categorical Features Association - Cramér\'s V')
plt.show()
```

# ● What are the top selling products in the superstore?

# Which category of products generates the highest revenue and profit?

```python
top_5_profit = (
    df.groupby('Product Name')['Profit']
    .sum()
    .nlargest(5)
)

display(top_5_profit)
```

```python
sns.barplot(
    x=top_5_profit.values,
    y=top_5_profit.index,
    palette='Set2'
)

plt.title('Top 5 Products by Profit')
plt.xlabel('Total Profit')
plt.ylabel('Product Name')
plt.show()
```

 # What is the impact of discounts and promotions on sales?

```python
sales_by_discount = (
    df.groupby('Discount')['Sales']
    .sum()
    .sort_values(ascending=False)
)

display(sales_by_discount)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=sales_by_discount.index,
    y=sales_by_discount.values,
    palette='Set2'
)

plt.title('Impact of Discounts on Sales')
plt.xlabel('Discount')
plt.ylabel('Total Sales')

plt.show()
```

```python
discount_analysis = df.groupby('Discount').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
}).reset_index()

display(discount_analysis)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

sns.barplot(
    data=discount_analysis,
    x='Discount',
    y='Sales',
    palette='Blues',
    ax=axes[0]
)

sns.barplot(
    data=discount_analysis,
    x='Discount',
    y='Profit',
    palette='Reds',
    ax=axes[1]
)

axes[0].set_title('Discount vs Sales')
axes[1].set_title('Discount vs Profit')

plt.tight_layout()
plt.show()
```
